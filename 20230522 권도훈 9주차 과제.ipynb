```
from pathlib import Path
import pandas as pd
from scipy.stats import chi2_contingency

DATA_DIR = Path('./')
file_path = DATA_DIR / 'channel_purchase.csv'

df = pd.read_csv(file_path)

print('[데이터 상위 5행]')
print(df.head(), '\n')

print('[변수 정보]')
print(df.info(), '\n')

table = pd.crosstab(df['channel'], df['purchase_yn'])

print('[교차표]')
print(table, '\n')

chi2, p, dof, expected = chi2_contingency(table)

print('[검정 결과]')
print('chi-square statistic:', round(chi2, 4))
print('p-value:', round(p, 6))
print('degrees of freedom:', dof)

expected_df = pd.DataFrame(
    expected,
    index=table.index,
    columns=table.columns
)

print('\n[기대도수]')
print(expected_df)

conversion_rate = (
    df.groupby('channel')['purchase_yn']
    .mean()
    .sort_values(ascending=False) * 100
)

print('\n[채널별 구매전환율(%)]')
print(conversion_rate.round(2))

print('\n[해석]')

if p < 0.05:
    print('유입채널과 구매여부는 서로 관련이 있습니다.')
else:
    print('유입채널과 구매여부의 관련성을 확인하기 어렵습니다.')

print(f'가장 높은 구매전환율 채널은 {conversion_rate.index[0]}입니다.')
print(f'전환율: {conversion_rate.iloc[0]:.2f}%')
```

```
from pathlib import Path
from scipy.stats import chi2_contingency

DATA_DIR = Path('./')
file_path = DATA_DIR / 'gender_preference.csv'

df = pd.read_csv(file_path)

print('[데이터 상위 5행]')
print(df.head(), '\n')

print('[변수 정보]')
print(df.info(), '\n')

table = pd.crosstab(df['gender'], df['content_type'])

print('[교차표]')
print(table, '\n')

chi2, p, dof, expected = chi2_contingency(table)

print('[검정 결과]')
print('chi-square statistic:', round(chi2, 4))
print('p-value:', round(p, 6))
print('degrees of freedom:', dof)

expected_df = pd.DataFrame(
    expected,
    index=table.index,
    columns=table.columns
)

print('\n[기대도수]')
print(expected_df)

top_pref = table.idxmax(axis=1)

print('\n[성별별 최다 선호 유형]')
print(top_pref)

row_ratio = (
    pd.crosstab(
        df['gender'],
        df['content_type'],
        normalize='index'
    ) * 100
)

print('\n[성별 내 콘텐츠 선호 비율(%)]')
print(row_ratio.round(2))

print('\n[해석]')

if p < 0.05:
    print('성별에 따라 콘텐츠 선호 분포가 다릅니다.')
else:
    print('성별에 따른 콘텐츠 선호 분포 차이를 확인하기 어렵습니다.')

for g in top_pref.index:
    print(f'{g}의 최다 선호 콘텐츠는 {top_pref[g]}입니다.')
```
