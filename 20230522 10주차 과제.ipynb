```
from pathlib import Path import pandas as pd from scipy import stats

DATA_DIR = Path('/content/drive/MyDrive/type3_week1') file_path = DATA_DIR / 'branch_satisfaction.csv'

try: df = pd.read_csv(file_path) except FileNotFoundError: print(f'파일을 찾을 수 없습니다: {file_path}') exit()

try: a = df[df['branch'] == 'A']['score'] b = df[df['branch'] == 'B']['score'] c = df[df['branch'] == 'C']['score'] except KeyError: print('CSV 파일에 branch 또는 score 컬럼이 없습니다.') exit()

result = stats.f_oneway(a, b, c) means = df.groupby('branch')['score'].mean().sort_values(ascending=False)

print(means) print('F-statistic:', round(result.statistic, 4)) print('p-value:', round(result.pvalue, 6)) print('평균이 가장 높은 지점:', means.index[0])
```

```
from pathlib import Path import pandas as pd import statsmodels.formula.api as smf from statsmodels.stats.anova import anova_lm

DATA_DIR = Path('/content/drive/MyDrive/type3_week1') file_path = DATA_DIR / 'ad_age_sales.csv'

try: df = pd.read_csv(file_path) except FileNotFoundError: print(f'파일을 찾을 수 없습니다: {file_path}') exit()

required_columns = ['sales_amount', 'ad_type', 'age_group'] for col in required_columns: if col not in df.columns: print(f'{col} 컬럼이 존재하지 않습니다.') exit()

model = smf.ols( 'sales_amount ~ C(ad_type) + C(age_group) + C(ad_type):C(age_group)', data=df ).fit()

anova_table = anova_lm(model, typ=2)

print(anova_table)

p_ad = anova_table.loc['C(ad_type)', 'PR(>F)'] p_age = anova_table.loc['C(age_group)', 'PR(>F)'] p_inter = anova_table.loc['C(ad_type):C(age_group)', 'PR(>F)']

print('광고유형 p-value:', round(p_ad, 6)) print('연령대 p-value:', round(p_age, 6)) print('상호작용 p-value:', round(p_inter, 6))
```
